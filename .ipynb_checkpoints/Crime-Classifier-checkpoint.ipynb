# Vancouver Crime Type Classifier

## Overview
This project applies K-Nearest Neighbours (KNN) classification to predict 
crime type based on historical incident data from the Vancouver Police 
Department. Model performance is evaluated on a held-out test set to ensure 
generalisation to unseen data.

## Data Source
- **Provider:** Vancouver Police Department (VPD) GeoDASH Open Data
- **Dataset:** Crime data from 2003 to present, updated weekly
- **URL:** https://geodash.vpd.ca/opendata/

## Features Used
| Feature | Description |
|---|---|
| `NEIGHBOURHOOD` | Vancouver neighbourhood where incident occurred |
| `HOUR` | Hour of day (0-23) |
| `MONTH` | Month of year (1-12) |
| `DAY_OF_WEEK` | Day of week derived from date |
| `SEASON` | Season derived from month |
| `YEAR` | Year of incident |

## Target Variable
`TYPE` — Crime category (e.g. Theft, Break & Enter, Assault, Mischief)

## Methodology
1. Data loading and exploration
2. Data cleaning
3. Train/test split
4. KNN model training
5. Model evaluation and results

## Data Preparation

The raw dataset contains separate YEAR, MONTH, and DAY columns. These are 
combined into a full date to extract DAY_OF_WEEK, since the same day number 
falls on different weekdays across years. DAY_OF_WEEK is a stronger predictor 
of crime type than the raw day number alone.

```r
crimes <- crimes |> 
    mutate(DATE = as_date(paste(YEAR, MONTH, DAY, sep = "-")),
          DAY_OF_WEEK = weekdays(DATE),
          TYPE = as_factor(TYPE),
          NEIGHBOURHOOD = as_factor(NEIGHBOURHOOD),
          DAY_OF_WEEK = as_factor(DAY_OF_WEEK)) |>
    select(TYPE, YEAR, MONTH, DAY, HOUR, NEIGHBOURHOOD, DATE, DAY_OF_WEEK)
head(crimes)
```

```r
crimes_clean <- crimes |>
    select(TYPE, YEAR, MONTH, HOUR, NEIGHBOURHOOD, DAY_OF_WEEK)
head(crimes_clean)
```

## Encoding Categorical Variables

KNN calculates distance between data points using numeric values, so categorical 
variables must be converted to numeric values. NEIGHBOURHOOD and 
DAY_OF_WEEK are encoded as numeric factors, where each unique category is 
assigned an integer. These are then scaled alongside the other numeric 
predictors.

```r
crimes_numeric <- crimes_clean |>
    mutate(NEIGHBOURHOOD = as.numeric(NEIGHBOURHOOD),
          DAY_OF_WEEK = as.numeric(DAY_OF_WEEK)) |>
    drop_na()
head(crimes_numeric)
Homs <- crimes_numeric |>
    filter(TYPE == "Vehicle Collision or Pedestrian Struck (with Fatality)") 
nrow(Homs)
```

```r
set.seed(123)
crimes_sample <- crimes_numeric %>%
  group_by(TYPE) %>%
  slice_sample(prop = 0.03) %>%
  ungroup()
crimes_sample
```

## Train/Test Split

The dataset is split into a training set (75%) and a test set (25%). 
Stratification by TYPE ensures that the proportion of each crime category 
is preserved in both sets. This prevents any crime type from being 
over or underrepresented.

```r
set.seed(1)
crimes_split <- initial_split(crimes_sample, prop = 0.75, strata = TYPE)
crimes_training <- training(crimes_split)
crimes_testing <- testing(crimes_split)
```

## Recipe

A recipe is created to define the model and preprocessing steps. 
TYPE is specified as the target variable, with NEIGHBOURHOOD, HOUR, MONTH, 
YEAR, and DAY_OF_WEEK as predictors. All numeric predictors are centred and 
scaled.

```r
crimes_recipe <- recipe(TYPE ~ YEAR	+ MONTH	+ HOUR + NEIGHBOURHOOD + DAY_OF_WEEK, data = crimes_training) |>
    step_scale(all_predictors()) |>
    step_center(all_predictors())
```

## Cross-Validation

The training data is split into 5 folds, stratified by TYPE, to tune the 
optimal value of K during model training.

```r
crimes_vfold <- vfold_cv(crimes_training, v = 5, strata = TYPE)
```

## Model Specification

Specifies a KNN classifier using the kknn engine, with the number of 
neighbours K left to be tuned during cross-validation

```r
crimes_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = tune()) |>
    set_engine("kknn") |>
    set_mode("classification")
```

## K Values

A table of K-values to perform Cross-Validation.

```r
k_val <- tibble(neighbors = seq(from = 1, to = 100, by = 5))
```

## Accuracies

The KNN model was tuned using v-fold cross-validation across a range of `k` values. 
For each value of `k`, the mean cross-validated accuracy was obtained using `tune_grid()` 
and collected using `collect_metrics()`.

> **Note:** A warning was raised during tuning indicating no observations were detected 
> for the `Homicide` crime type in some cross-validation folds. This is expected given 
> how rare homicide incidents are in the dataset. Those levels were ignored during metric 
> computation.

The `accuracies` table filters the tuning results to only the accuracy metric, and 
`accuracy_plot` visualizes mean accuracy vs. number of neighbors, helping identify 
the optimal `k` value before final model evaluation on the test set.

```r
knn_results <- workflow() |>
    add_recipe(crimes_recipe) |>
    add_model(crimes_spec) |>
    tune_grid(resamples = crimes_vfold, grid = k_val) |>
    collect_metrics()
```

```r
accuracies <- knn_results |>
    filter(.metric == "accuracy")
```

```r
accuracy_plot <- ggplot(accuracies, aes(x = neighbors, y = mean)) +
    geom_point() +
    geom_line() +
    labs(x = "Neighbors", y = "Accuracy") +
    theme(text = element_text(size = 12))
accuracy_plot
```

```r
optimal_k <- accuracies |>
    arrange(desc(mean)) |>
    head(1) |>
    pull(neighbors)
optimal_k
```

## Model Specification with Optimal K 

Based on the accuracy plot, `k = 96` was found to be the optimal number of neighbours. 
The model was specified using a rectangular weight function and fit on the full 
training set using this value.

```r
crimes_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = 96) |>
    set_engine("kknn") |>
    set_mode("classification")
crimes_fit <- workflow() |>
    add_recipe(crimes_recipe) |>
    add_model(crimes_spec) |>
    fit(data = crimes_training)
```

## Predicting Labels in the testing set


The fitted model was used to generate predictions on the held-out test set.

```r
crime_predictions <- predict(crimes_fit, crimes_testing) |>
    bind_cols(crimes_testing)
crime_predictions
```

## Measures of Accuracy

```r
confusion_matrix <- crime_predictions |>
    conf_mat(truth = TYPE, estimate = .pred_class)
confusion_matrix |>
  autoplot(type = "heatmap")+
  theme(axis.text.x = element_text(angle = 45, hjust = 1))
```

## Discussion

The confusion matrix reveals that the model performs best on the most common 
crime types. Other Theft and Theft from Vehicle show the strongest correct 
predictions, shown by the darkest cells along the diagonal. Offence 
Against a Person is also predicted with reasonable levels of accuracy.

However, the model struggles with rare categories. Homicide, Theft of Bicycle, 
Theft of Vehicle, and Vehicle Collision are never predicted correctly. This is due the the fact that the such categories are rare in nature, being unable to provide the model with enough instances of these crimes to learn their 
patterns. Additionally, Break and Enter Commercial is frequently misclassified as Theft 
from Vehicle or Other Theft.

This reflects a challenge known as class imbalance, where 
the model becomes biased toward predicting frequent categories. Next steps would involve apply oversampling techniques to rare crime types to improve predictions across these categories.


```r
crime_predictions |> 
    metrics(truth = TYPE, estimate = .pred_class) |>
    filter(.metric == "accuracy")
```

## Results

The KNN classifier achieved an accuracy of 38.4% on the test set.

Given 11 crime classes, a random classifier would achieve roughly ~9% accuracy. However, 38.4% is still quite low 
for a practical classifier, which is expected given the features used (i.e., time of day, 
month, day of week, season, year, and neighbourhood). These variables alone are not always strong predictors for 
distinguishing between crime types. Many crimes occur across all times and 
neighbourhoods without a strong, distinguishable pattern. Additionally, incident-level features would 
likely improve the performance of the classifier.
